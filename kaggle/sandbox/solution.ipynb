```python
# === Step 0: Imports & display settings ===
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import RobustScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn import set_config

set_config(transform_output="pandas")
pd.set_option("display.max_columns", 120)
pd.set_option("display.width", 180)
print("Libraries imported.")
```

```python
# === Step 1: Load data ===
# Make sure the file 'Ames_Housing_Data.csv' is in the same directory as this notebook.
# If your file is elsewhere, update the path accordingly.
csv_path = "train.csv"
df = pd.read_csv(csv_path)
print("Data loaded:", df.shape)
df.head(10)
```

```python
# === Step 2: Identify numeric and textual columns ===
num_col = df.select_dtypes(include=["number"])
text_col = df.select_dtypes(include=["object"])

print(num_col.head(10))
print('-------------------------------')
print(text_col.head(10))
```

```python
# dealing missing data
# scal(num collumns) & encoding (text collumns)
num_pipe = Pipeline(steps=[
    ('impute', SimpleImputer(strategy='median')),
    ('scale', RobustScaler())  # ทน outlier
])

cat_pipe = Pipeline(steps=[
    ('impute', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=True))
])
```

```python
preprocess = ColumnTransformer(
    transformers=[
        ('num', num_pipe, num_col),
        ('cat', cat_pipe, text_col)
    ],
    remainder='drop'
)

preprocess
```
